```python

import numpy as np
import pandas as pd
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Flatten, Dense
from tensorflow.keras.utils import to_categorical
from sklearn.preprocessing import LabelEncoder

# 1. Load Dataset
data = pd.read_csv("D:\data science\DL-lab\lab\data2.csv")

sentences = data["sentence"].astype(str).tolist()
categories = data["category"].astype(str).tolist()

# 2. Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)

total_words = len(tokenizer.word_index) + 1
sequences = tokenizer.texts_to_sequences(sentences)

max_length = max(len(seq) for seq in sequences)
X = pad_sequences(sequences, maxlen=max_length, padding="post")

# 3. Encode Labels
label_encoder = LabelEncoder()
y_int = label_encoder.fit_transform(categories)
y = to_categorical(y_int)

num_classes = y.shape[1]

# 4. Build Model
model = Sequential()
model.add(Embedding(total_words, 16, input_length=max_length))
model.add(Flatten())
model.add(Dense(16, activation="relu"))
model.add(Dense(num_classes, activation="softmax"))

# 5. Compile Model
model.compile(
    loss="categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

# 6. Train Model
model.fit(X, y, epochs=200, verbose=0)

# 7. Classify New Text
def classify(text):
    seq = tokenizer.texts_to_sequences([text])
    padded = pad_sequences(seq, maxlen=max_length, padding="post")

    prediction = model.predict(padded, verbose=0)
    class_index = np.argmax(prediction)

    return label_encoder.inverse_transform([class_index])[0]

# 8. Test the Model
sample = input("Enter a sentence: ")
print("Predicted Category:", classify(sample))
```

```python

import numpy as np
import pandas as pd
import random
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GlobalAveragePooling1D, Dense
from tensorflow.keras.utils import to_categorical
from sklearn.preprocessing import LabelEncoder

# 1. Load Dataset
df = pd.read_csv("D:\data science\DL-lab\lab\data1.csv")

patterns = df["pattern"].astype(str).tolist()
labels = df["intent"].astype(str).tolist()

# 2. Canned Responses
responses = {
    "greeting": ["Hello! How can I help you today?"],
    "goodbye": ["Goodbye! Have a great day."],
    "thanks": ["You're welcome!"],
    "complaint": ["I'm sorry to hear that. Let me help resolve this."],
    "request_help": ["Sure, I can help you with that."],
    "order_status": ["Let me check that for you right away."],
    "book_appointment": ["Sure, let's get that scheduled for you."],
    "weather_query": ["Let me check the forecast for you."]
}

# 3. Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts(patterns)

total_words = len(tokenizer.word_index) + 1
sequences = tokenizer.texts_to_sequences(patterns)

max_length = max(len(seq) for seq in sequences)
X = pad_sequences(sequences, maxlen=max_length, padding="post")

# 4. Encode Labels
label_encoder = LabelEncoder()
y_int = label_encoder.fit_transform(labels)
y = to_categorical(y_int)
num_classes = y.shape[1]

# 5. Build Neural Network
model = Sequential()
model.add(Embedding(total_words, 8, input_length=max_length))
model.add(GlobalAveragePooling1D())
model.add(Dense(16, activation="relu"))
model.add(Dense(num_classes, activation="softmax"))

# 6. Compile Model
model.compile(
    loss="categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

# 7. Train Model
model.fit(X, y, epochs=200, verbose=0)

# 8. Generate Dialogue Response
def get_response(text):
    seq = tokenizer.texts_to_sequences([text])
    padded = pad_sequences(seq, maxlen=max_length, padding="post")

    prediction = model.predict(padded, verbose=0)
    intent = label_encoder.inverse_transform(
        [np.argmax(prediction, axis=1)[0]]
    )[0]

    if intent in responses:
        return random.choice(responses[intent])
    else:
        return "Sorry, I don't understand."

# 9. Test the Dialogue System
user_input = input("You: ")
print("Bot:", get_response(user_input))
```

```python

import pandas as pd
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.optimizers import Adam

# 1. Load Dataset
data = pd.read_csv("D:\data science\DL-lab\lab\LAB_8.csv")

text = data["sentence"].astype(str).tolist()

# 2. Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts(text)
total_words = len(tokenizer.word_index) + 1

# 3. Create Input Sequences
sequences = []

for sentence in text:
    token = tokenizer.texts_to_sequences([sentence])[0]
    for i in range(1, len(token)):
        sequences.append(token[:i + 1])

# 4. Padding
max_length = max(len(seq) for seq in sequences)
sequences = pad_sequences(sequences, maxlen=max_length, padding="pre")

# 5. Split Input and Output
X = sequences[:, :-1]
y = sequences[:, -1]
y = to_categorical(y, num_classes=total_words)

# 6. Build Model
model = Sequential()
model.add(Embedding(input_dim=total_words, output_dim=32))
model.add(LSTM(64))
model.add(Dense(total_words, activation="softmax"))

# 7. Compile Model
optimizer = Adam(learning_rate=0.001, clipnorm=1.0)

model.compile(
    loss="categorical_crossentropy",
    optimizer=optimizer,
    metrics=["accuracy"]
)

# 8. Train Model
model.fit(X, y, epochs=100, verbose=1)

# 9. Generate Text
seed = input("Enter Starting Text: ")

for i in range(5):
    token = tokenizer.texts_to_sequences([seed])[0]
    token = pad_sequences([token], maxlen=max_length - 1, padding="pre")

    prediction = np.argmax(model.predict(token, verbose=0), axis=-1)[0]

    output_word = ""
    for word, index in tokenizer.word_index.items():
        if index == prediction:
            output_word = word
            break

    if output_word == "":
        break

    seed = seed + " " + output_word

print("Generated Text:", seed)
```

```python

import numpy as np
import pandas as pd
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GlobalAveragePooling1D, Dense
from tensorflow.keras.regularizers import l1, l2

# 1. Load Dataset
df = pd.read_csv("D:\data science\DL-lab\lab\data5.csv")

sentences = df["sentence"].astype(str).tolist()
labels = df["label"].astype(int).tolist()

# 2. Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)

total_words = len(tokenizer.word_index) + 1
sequences = tokenizer.texts_to_sequences(sentences)

max_length = max(len(seq) for seq in sequences)
X = pad_sequences(sequences, maxlen=max_length, padding="post")
y = np.array(labels)

embedding_dim = 8

# 3. Build Model
model = Sequential()
model.add(Embedding(
    total_words,
    embedding_dim,
    embeddings_regularizer=l2(0.001)
))
model.add(GlobalAveragePooling1D())
model.add(Dense(8, activation="relu", kernel_regularizer=l1(0.001)))
model.add(Dense(1, activation="sigmoid"))

# 4. Compile Model
model.compile(
    loss="binary_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

# 5. Train Model
history = model.fit(X, y, epochs=200, verbose=0)

print(f"Final Training Loss: {history.history['loss'][-1]:.4f}")
print(f"Final Training Accuracy: {history.history['accuracy'][-1]:.4f}")

# 6. Create Sentence Embedding Model
sentence_embedding_model = Sequential(model.layers[:2])

# 7. Convert Input Text into Numbers
sample = input("Enter a sentence: ")

seq = tokenizer.texts_to_sequences([sample])
padded = pad_sequences(seq, maxlen=max_length, padding="post")

# 8. Get Sentence Embedding
embedding = sentence_embedding_model.predict(padded, verbose=0)[0]
print("Sentence Embedding:", embedding)

# 9. Predict Sentiment
prediction = model.predict(padded, verbose=0)[0][0]

if prediction >= 0.5:
    print("Predicted Sentiment: Positive")
else:
    print("Predicted Sentiment: Negative")
```

```python
# 1. Import Libraries
import numpy as np
import pandas as pd
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GlobalAveragePooling1D, Dense
from tensorflow.keras.regularizers import l1, l2

# 2. Read Dataset
df = pd.read_csv("D:\data science\DL-lab\lab\data5.csv")

sentences = df["sentence"].astype(str).tolist()
labels = df["label"].astype(int).tolist()

# 3. Tokenization and Padding
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)

total_words = len(tokenizer.word_index) + 1
sequences = tokenizer.texts_to_sequences(sentences)
max_length = max(len(seq) for seq in sequences)

X = pad_sequences(sequences, maxlen=max_length, padding="post")
y = np.array(labels)

embedding_dim = 8

# 4. Build Model with L1 and L2 Regularization
model = Sequential()
model.add(Embedding(total_words, embedding_dim,
                    embeddings_regularizer=l2(0.001)))
model.add(GlobalAveragePooling1D())
model.add(Dense(8, activation="relu",
                kernel_regularizer=l1(0.001)))
model.add(Dense(1, activation="sigmoid"))

# 5. Compile Model
model.compile(loss="binary_crossentropy",
              optimizer="adam",
              metrics=["accuracy"])

# 6. Train Model
history = model.fit(X, y, epochs=200, verbose=0)

print("Final Training Loss:",
      round(history.history["loss"][-1], 4))
print("Final Training Accuracy:",
      round(history.history["accuracy"][-1], 4))

# 7. Create Sentence Embedding Extractor
sentence_embedding_model = Sequential(model.layers[:2])

# 8. Function to Get Sentence Embedding
def get_sentence_embedding(text):
    seq = tokenizer.texts_to_sequences([text])
    padded = pad_sequences(seq, maxlen=max_length, padding="post")
    return sentence_embedding_model.predict(padded, verbose=0)[0]

# 9. Function to Classify Sentiment
def classify(text):
    seq = tokenizer.texts_to_sequences([text])
    padded = pad_sequences(seq, maxlen=max_length, padding="post")
    prediction = model.predict(padded, verbose=0)[0][0]
    return "Positive" if prediction >= 0.5 else "Negative"

# 10. Test the Model
sample = input("Enter a sentence: ")

print("Sentence Embedding:", get_sentence_embedding(sample))
print("Predicted Sentiment:", classify(sample))
```
